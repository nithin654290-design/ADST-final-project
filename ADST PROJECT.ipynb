```python
import pandas as pd

profiles = pd.read_csv("data/profiles.csv")
study_plans = pd.read_csv("data/study_plans.csv")
user_roles = pd.read_csv("data/user_roles.csv")

print("Profiles shape:", profiles.shape)
print("Study plans shape:", study_plans.shape)
print("User roles shape:", user_roles.shape)
```

```python
print("PROFILES COLUMNS:")
print(profiles.columns.tolist())

print("\nSTUDY PLANS COLUMNS:")
print(study_plans.columns.tolist())

print("\nUSER ROLES COLUMNS:")
print(user_roles.columns.tolist())
```

```python

# Check common column names across the datasets
common_keys = set(profiles.columns) & set(study_plans.columns)
merge_key = list(common_keys)[0] if common_keys else "user_id"

# Merge profiles with study_plans
merged_data = profiles.merge(
    study_plans,
    on=merge_key,
    how="left"
)

# Merge with user_roles if applicable
if merge_key in user_roles.columns:
    merged_data = merged_data.merge(
        user_roles,
        on=merge_key,
        how="left"
    )

print("All datasets merged successfully!")
print("Final dataset shape:", merged_data.shape)
merged_data.head()
```

```python
merged_data.to_csv("data/merged_dataset.csv", index=False)
print("Final dataset saved successfully!")
```
